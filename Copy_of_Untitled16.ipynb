```python
# Try except (error handling)
try:
  result = 10 / 0
  print("this line will be executed.")
```

```python
except ZeroDivisionError:
  print("You can't divide by zero.")
```

```python
print("The program continues to run smoothly")
```

```python
# fainally always run whatever happens
try:
  # This code triggers a Zero division error
  result = 10 / 0
except ZeroDivisionError:
  # This block handles the exception
  print("error: You can not devide by zero!")
else:
  # This block is skipped completely
  print(f"Success! The answer is {result}")

finally:
  # Always run no matter what happens above
  print("Cleanup: This line always executes.")
```
